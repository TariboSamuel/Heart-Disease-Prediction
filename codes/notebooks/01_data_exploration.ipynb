### **Target Variable Analysis: Class Distribution**

Before modeling, it is critical to evaluate the distribution of the target variable (`TenYearCHD`) to determine if there is a class imbalance. A severe imbalance can cause models to become biased toward the majority class.


```python
# Returns the unique values and their counts in the target variable "TenYearCHD"
df["TenYearCHD"].value_counts()
```

The output of the `.value_coiunts()` above confirms a significant class imbalance (3,596 negative cases vs. 644 positive cases). This indicates that the dataset is highly skewed toward patients without heart disease. Consequently, resampling techniques (such as SMOTE-ENN) and algorithm-level interventions (like Class Weights) must be implemented during model training to ensure the minority class is properly learned.

### **Target Variable Analysis: Feature Correlation**

Subsequently, the Pearson correlation coefficients between all independent physiological features and the target variable were calculated. This analysis was conducted to identify the primary linear risk factors for cardiovascular disease.

```python
# Returns the correlation of all features with the target variable "TenYearCHD" and sorts them in descending order
df.corr()["TenYearCHD"].sort_values(ascending=False)
```

The features demonstrating the strongest positive linear correlation to heart disease risk are `age` (0.225), `sysBP` (0.216), and `prevalentHyp` (0.177). Because these primary clinical risk factors possess a clear linear relationship with the target variable, linear algorithms (such as Logistic Regression) are theoretically well-suited for this dataset. Consequently, it is hypothesized that linear models may perform competitively against complex non-linear ensembles (such as Random Forest or XGBoost).
