# **2. Data Cleaning & Preprocessing**

This stage prepares the raw dataset for analysis and machine learning by addressing data quality issues such as missing values, duplicate records, and inconsistent data. The process begins by loading the dataset and importing the required libraries.

### **Library Imports**

The following libraries are imported to support data manipulation, numerical computations, and data visualization throughout the preprocessing and analysis stages.

```python
# Importing Libraries

import pandas as pd          # For data manipulation
import numpy as np            # For mathematical operation
import matplotlib.pyplot as plt    # For data visualization
import seaborn as sns              # For data visualization
```

## **Data Ingestion**

In this section, the raw Framingham dataset is loaded into the working environment and copied into memory, thereby preparing it for cleaning.


```python
# Loading the Framingham dataset
framingham_df = pd.read_csv("../data/raw/framingham.csv")

df = framingham_df.copy()  # Creating a copy of the original dataset to work with   

print('Framingham Data Loaded Successfully: \n')

print("The first 5 rows of the Framingham data: \n")
df.head()
```

### **Handling Missing Values**

In healthcare datasets, missing values must be handled carefully to avoid introducing bias or data leakage. Given the continuous nature of the physiological variables in this dataset, missing values were imputed using the median of each respective feature. 
Median imputation was selected over mean imputation because it is highly robust to outliers (e.g., extremely high glucose or cholesterol readings), ensuring that the central tendency of the data is not artificially skewed.

```python
# Impute missing values using the median
numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Verify imputation
print("Missing values after imputation:\n")
df.isnull().sum()
```


The output confirms that all missing values have been successfully imputed, leaving no missing observations in the dataset.

### **Data Type Formatting**

After imputing missing values, some categorical and discrete variables were automatically converted to floating-point data types. To maintain data consistency and preserve their intended representation, these variables were cast back to integer data types.

```python
# Cast appropriate columns to integer
cols_to_int = ['education', 'cigsPerDay', 'BPMeds', 'prevalentStroke', 'prevalentHyp', 'diabetes']
df[cols_to_int] = df[cols_to_int].astype(int)

# Verify data types
df.info()
```

The output confirms that the selected variables have been successfully converted to their appropriate data types, ensuring consistency for subsequent analysis.

### **Exporting Cleaned Data**

The cleaned dataset is exported to the processed data directory for use in subsequent stages, including feature engineering and model development.


```python
# Export the cleaned dataframe to a new CSV file
df.to_csv("../data/processed/framingham_cleaned.csv", index=False)

print("Cleaned dataset successfully exported to 'data/processed/framingham_cleaned.csv'.")
```

The dataset has now been fully preprocessed and is ready for exploratory data analysis, feature engineering, and machine learning model development.
